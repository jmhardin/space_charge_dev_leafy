```julia
using Random
using BeamTracking
using CairoMakie;
using Statistics

function my_std(a)
    n = length(a)
    sum = 0
    sum_sq = 0

    for e in a
        sum = sum + e
        sum_sq = sum_sq + e*e
    end

    std_dev = sqrt(sum_sq/(n) - sum*sum/((n)*(n)))
    return std_dev
end
print(my_std([1,2,3,4]))

N_arr = [10,100,1000,10000]

function x_space(N,plot=true)
    p0c = 10e6
    p_over_q_ref = BeamTracking.E_to_R(Species("electron"), sqrt(p0c^2 + massof(Species("electron"))^2))
    Random.seed!(1)
    
    drift = Drift(L = 2.0, 
    tracking_method = Symplectic(ibs_damping_on = true, ibs_fluctuations_on = true))
    line = Beamline([drift], species_ref = Species("electron"), E_ref = 18e9)
    
    v0 = randn(N,6)
    #println(v0)
    #1 um std dev in x and y
    v0[:,XI] = v0[:,XI]/1e6
    v0[:,YI] = v0[:,YI]/2e6
    #1 mm in z
    v0[:,ZI] = v0[:,ZI]/1e3

    #100 MeV electrons with 1 MeV spreaed
    v0[:,PZI] = v0[:,PZI]*1e6 + ones(N,1)*1e8;
    v0[:,PXI] = v0[:,PXI]*1e3
    v0[:,PYI] = v0[:,PYI]*1e3

    
    arr = deepcopy(v0[:,PXI])
    #println("%12.02f",my_std(arr));
    b0 = Bunch(copy(v0), species = line.species_ref, p_over_q_ref = line.p_over_q_ref)
    track!(b0, line; use_explicit_SIMD = !(VERSION < v"1.11" && Sys.ARCH == :x86_64))
    
    # Defining the arrays
    #arr = v0[:,PXI];
    arr_2 = b0.v[:,PXI];
    if (plot)
        scatter(arr, arr_2; axis = (; title = string(N),xlabel="X_in",ylabel="X_out"))
    end
    return my_std(arr),my_std(arr_2)
end
```

```julia
using Printf
using FFTW
@inline faddeeva(z) = faddeeva(z, Val(48))

function __faddeeva_coeff(n, type::Type=Float64)
  N = n
  M = 2*N
  L = sqrt(N/sqrt(type(2)))
  theta_arr = [k*type(pi)/M for k in (-M+1):(M-1)]
  t_arr = L*tan.(theta_arr/2)
  f_arr = [(L^2 + t^2)*exp(-t^2) for t in t_arr]
  f_arr = vcat(zero(type), f_arr)
  a_arr = real.(fft(fftshift(f_arr)))/(2*M)
  return a_arr[2:(N+1)], L
end

@generated function faddeeva(z::Complex{T}, ::Val{N}) where {T,N}
  a_arr, L = __faddeeva_coeff(N, T)
  a_arr = reverse(a_arr)
  ex = :($(a_arr[1]))
  for i in 2:N
    ex = :(muladd($ex, Z, $(a_arr[i])))
  end
  return quote
    c1 = imag(z) >= 0 ? 0 : 1
    c2 = 1 - 2*c1
    z = c2*z
    lmiz_inv = 1/($L - im*z)
    Z = ($L + im*z)*lmiz_inv
    return c1*2*exp(-z^2) + c2*(2*$ex*lmiz_inv + $(1/sqrt(T(pi))))*lmiz_inv
  end
end

@inline function basetti_erskin_E(x,y,sigma_x,sigma_y)
    sx2 = sigma_x*sigma_x
    sy2 = sigma_y*sigma_y
    x2 = x*x
    y2 = y*y
    sig_denom = Complex(sx2-sy2)
    A = sqrt(Complex(2*pi/sig_denom))
    B = faddeeva(Complex(x,y)/sqrt(2*sig_denom))
    C = faddeeva(Complex(x*sigma_y/sigma_x,y*sigma_x/sigma_y)/sqrt(2*sig_denom))
    expD = exp(-x2/(2*sx2) - y2/(2*sy2))
    result = A*(B - expD*C)
    return result
end;

@inline function basetti_erskin_piecewise(x,y,sigma_x,sigma_y)
    if (sigma_x > sigma_y)
        out = @inline basetti_erskin_E(x,y,sigma_x,sigma_y)
        @printf("dx = %8.02e dy = %8.02e\n", imag(out),real(out))
    elseif (sigma_x < sigma_y)
        out = @inline basetti_erskin_E(y,x,sigma_y,sigma_x)
        @printf("dx = %8.02e dy = %8.02e\n", real(out),imag(out))
    end
end
```

```julia
@inline function basetti_erskin_w(z)
    return exp(-z*z)*(1-erf(Complex(0,-1)*z))
end

@inline function basetti_erskin_PI(x,y,sigma_x,sigma_y)
    sx2 = sigma_x*sigma_x
    sy2 = sigma_y*sigma_y
    x2 = x*x
    y2 = y*y
    sig_denom = Complex(sx2-sy2)
    A = sqrt(Complex(2*pi/sig_denom))
    B = basetti_erskin_w(Complex(x,y)/sqrt(2*sig_denom))
    C = basetti_erskin_w(Complex(x*sigma_y/sigma_x,y*sigma_x/sigma_y)/sqrt(2*sig_denom))
    expD = exp(-x2/(2*sx2) - y2/(2*sy2))
    result = A*(B - expD*C)
    return result
end;

@inline function basetti_erskin_piecewise(x,y,sigma_x,sigma_y)
    if (sigma_x > sigma_y && y > 0)
        out = @inline basetti_erskin_PI(x,y,sigma_x,sigma_y)
        @printf("dx = %8.02e dy = %8.02e\n", real(out),imag(out))
    elseif (sigma_x > sigma_y && y < 0)
        out = @inline basetti_erskin_PI(x,-y,sigma_x,sigma_y)
        @printf("dx = %8.02e dy = %8.02e\n", -real(out),imag(out))
    elseif (sigma_x < sigma_y && x > 0)
        out = @inline basetti_erskin_PI(y,x,sigma_y,sigma_x)
        @printf("dx = %8.02e dy = %8.02e\n", imag(out),real(out))
    elseif (sigma_x < sigma_y && x < 0)
        out = @inline basetti_erskin_PI(y,-x,sigma_y,sigma_x)
        @printf("dx = %8.02e dy = %8.02e\n", imag(out),-real(out))
    end
end
```

```julia
"""
    wofz_branchfree(xi, yi; n_terms=32, nu_terms=48) -> (u, v, overflow)
 
Branch-free variant of `wofz` (Poppe & Wijers, ACM TOMS Algorithm 680):
every code path uses `ifelse` (select/cmov) instead of `if`/`else`, so
there is no data-dependent control-flow divergence. This is intended for
use inside `@simd` loops, broadcasts over arrays, or GPU kernels, where
branching on `qrho` per-element would otherwise serialize/diverge.
 
--- Trade-offs vs. the original adaptive `wofz` ---
 
1. All three formulas (power series / continued fraction / Taylor with
   continued-fraction derivatives) are evaluated IN FULL on every call,
   then blended with `ifelse`. The original only ever executes one.
   This roughly doubles to quintuples the flop count per call.
 
2. The adaptive term counts `N`/`NU`/`KAPN` in the original exist
   specifically to do the *minimum* work needed for ~14-digit accuracy
   given the input. Making the loop trip count data-independent means
   giving up that adaptivity: `n_terms` and `nu_terms` below are FIXED,
   sized to the worst case that occurs anywhere in the region where
   each formula is actually selected. The defaults (32, 48) are chosen
   conservatively; you can raise them if you see accuracy loss near the
   region boundaries (qrho ≈ 0.085264 and qrho ≈ 1), or lower them for
   speed if your inputs stay well inside one region.
 
3. `KAPN` (which terms of the Taylor sum to keep) is still genuinely
   input-dependent — that's not a loop-trip-count optimization, it's
   part of the math — so it survives as a continuous mask
   (`n <= kapn`) rather than an integer early-exit.
 
4. Overflow is reported via a boolean mask instead of an early return
   (there's no `GOTO 100` equivalent). On overflow, `u`/`v` may contain
   `Inf`/`NaN` and the caller should check/discard using the returned
   flag.
 
I have not been able to execute this in this environment (no Julia
runtime available here) — validate it against the original `wofz`
over your input domain, especially near the two region boundaries,
before relying on it.
 
Reference: G.P.M. Poppe, C.M.J. Wijers, "More Efficient Computation of
the Complex Error Function", ACM Trans. Math. Software 16 (1990).
"""
function wofz_branchfree(xi::T, yi::T; n_terms::Int=32, nu_terms::Int=48) where {T<:AbstractFloat}
    FACTOR   = T(1.12837916709551257388)   # 2/sqrt(pi)
    RMAXREAL = T(0.5e154)
    RMAXEXP  = T(708.503061461606)
    RMAXGONI = T(3.53711887601422e15)
 
    xabs = abs(xi)
    yabs = abs(yi)
 
    # would-overflow mask; clamp downstream inputs so the rest of the
    # arithmetic runs unconditionally without producing spurious Inf/NaN
    overflow1 = (xabs > RMAXREAL) | (yabs > RMAXREAL)
    xabsS = ifelse(overflow1, zero(T), xabs)
    yabsS = ifelse(overflow1, zero(T), yabs)
 
    x = xabsS / T(6.3)
    y = yabsS / T(4.4)
    qrho  = x^2 + y^2
    xquad = xabsS^2 - yabsS^2
    yquad = 2 * xabsS * yabsS
 
    regionA = qrho < T(0.085264)
    regionC = qrho > one(T)
 
    # ---------------- region A: power series (fixed term count) ----------------
    j = T(2*n_terms + 1)
    xsum = one(T) / j
    ysum = zero(T)
    for i in n_terms:-1:1
        j -= 2
        xaux = (xsum*xquad - ysum*yquad) / i
        ysum = (xsum*yquad + ysum*xquad) / i
        xsum = xaux + one(T)/j
    end
    u1 = -FACTOR*(xsum*yabsS + ysum*xabsS) + one(T)
    v1 =  FACTOR*(xsum*xabsS - ysum*yabsS)
    daux = exp(-xquad)
    u2A =  daux*cos(yquad)
    v2A = -daux*sin(yquad)
    uA = u1*u2A - v1*v2A
    vA = u1*v2A + v1*u2A
 
    # ------- regions B/C: continued fraction / Taylor (fixed term count) -------
    qrhoB = (1 - y) * sqrt(max(1 - qrho, zero(T)))
    h     = ifelse(regionC, zero(T), T(1.88)*qrhoB)
    h2    = ifelse(h > 0, 2*h, one(T))                  # dummy keeps h2^kapn finite when h==0
    kapn  = round(ifelse(regionC, T(-1), T(7) + T(34)*qrhoB))
    b     = h > 0
 
    qlambda = ifelse(b, h2^kapn, zero(T))
 
    rx = zero(T); ry = zero(T); sx = zero(T); sy = zero(T)
    for n in nu_terms:-1:0
        np1 = T(n + 1)
        tx = yabsS + h + np1*rx
        ty = xabsS - np1*ry
        c  = T(0.5) / (tx^2 + ty^2)
        rx = c*tx
        ry = c*ty
        keep = b & (n <= kapn)
        txl  = qlambda + sx
        sx = ifelse(keep, rx*txl - ry*sy, sx)
        sy = ifelse(keep, ry*txl + rx*sy, sy)
        qlambda = ifelse(keep, qlambda/h2, qlambda)
    end
 
    uBC = ifelse(h == 0, FACTOR*rx, FACTOR*sx)
    vBC = ifelse(h == 0, FACTOR*ry, FACTOR*sy)
    uBC = ifelse(yabsS == 0, exp(-xabsS^2), uBC)
 
    u = ifelse(regionA, uA, uBC)
    v = ifelse(regionA, vA, vBC)
 
    # ---------------- quadrant correction ----------------
    isNeg     = yi < 0
    xquadN    = -xquad
    overflow2 = (yquad > RMAXGONI) | (xquadN > RMAXEXP)
    w1        = 2*exp(ifelse(overflow2, zero(T), xquadN))
    u2alt     =  w1*cos(yquad)
    v2alt     = -w1*sin(yquad)
 
    u2f = ifelse(regionA, 2*u2A, u2alt)
    v2f = ifelse(regionA, 2*v2A, v2alt)
 
    uNeg = u2f - u
    vNeg = v2f - v
 
    uOut = ifelse(isNeg, uNeg, u)
    vSel = ifelse(isNeg, vNeg, v)
    flipV = ifelse(isNeg, xi > 0, xi < 0)
    vOut = ifelse(flipV, -vSel, vSel)
 
    # overflow2 only actually applies on the non-power-series path, matching
    # the original (it's only computed inside that ELSE branch there)
    #overflow = overflow1 | (isNeg & !regionA & overflow2)
 
    return uOut, vOut
end
 
"""
    wofz_branchfree(z::Complex; kwargs...)
 
Convenience wrapper returning a `Complex` result. Still purely real
arithmetic internally.
"""
function wofz_branchfree(z::Complex{T}; kwargs...) where {T<:AbstractFloat}
    u, v = wofz_branchfree(real(z), imag(z); kwargs...)
    return Complex(u, v)
end
```
